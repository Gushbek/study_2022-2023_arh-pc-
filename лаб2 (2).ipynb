- Построил гистограммы плотности и наложил теоретическую кривую. При n = 1000 гистограмма почти совпадает с теорией, при n = 100 сильно изрезана. Плотность, как и CDF, сходится к теоретической с ростом n.



```python
from scipy import stats

def fun_norm_pdf(x):
    return stats.norm.pdf(x, 0, 1);
```

```python
theory_norm_x=np.linspace(-4,4,100)
theory_norm_y=fun_norm_pdf(theory_norm_x)
```

```python
plt.figure(dpi=100)
plt.title('Плотность распределения (нормальное, N(0,1))')
plt.plot(theory_norm_x,theory_norm_y, label='theory')
plt.hist(norm_100,  density=True, label='norm:100')
plt.hist(norm_1000, density=True, label='norm:1000')
plt.legend()
plt.show()
```

- То же для нормального: при n = 1000 плотность гладкая и близка к теории, при n = 100 — много случайных отклонений. Вывод — большая выборка даёт более точную оценку плотности.

```python
def exp_error(n):
    data=np.random.exponential(1, size=n)
    bins=np.linspace(0,10,30)
    emp,_=np.histogram(data, bins=bins, density=True)
    centers=(bins[:-1]+bins[1:])/2
    theor=fun_exp_pdf(1,centers)
    return np.mean(np.abs(emp-theor))
```

```python
def norm_error(n):
    data=np.random.normal(0,1,size=n)
    bins=np.linspace(-4,4,30)
    emp,_=np.histogram(data, bins=bins, density=True)
    centers=(bins[:-1]+bins[1:])/2
    theor=fun_norm_pdf(centers)
    return np.mean(np.abs(emp-theor))
    
```

```python
sizes=[10, 50, 100, 500, 1000, 5000, 10000]
err_exp =[exp_error(n)  for n in sizes]
err_norm=[norm_error(n) for n in sizes]

for n,e1,e2 in zip(sizes, err_exp, err_norm):
    print(f"n={n:<6} exp MAE = {e1:}   norm MAE = {e2:}")
```

- Я посчитал MAE между эмпирической и теоретической плотностью для разных n. Ошибка уменьшается с ростом n — примерно как 1/√n. Значит, точность аппроксимации растёт с количеством экспериментов

## Задание 2 

```python
def dist(x1, y1, x2, y2):
    return np.sqrt((x2-x1)**2 + (y2-y1)**2)

def distance_line(number):
    a = 10
    b = 30
    distance = []
    for i in range(number):
        x1 = np.random.uniform(0, a)
        y1 = np.random.uniform(0, b)

        x2 = np.random.uniform(0, a)
        y2 = np.random.uniform(0, b)

        distance.append(dist(x1, y1, x2, y2))

    distance.sort()
    return distance
```

```python
dist_100   = np.array(distance_line(100))
dist_1000  = np.array(distance_line(1000))
dist_10000 = np.array(distance_line(10000))
```

```python
print('Среднее расстояние для n=100   =', np.average(dist_100))
print('Среднее расстояние для n=1000  =', np.average(dist_1000))
print('Среднее расстояние для n=10000 =', np.average(dist_10000))
```

```python
dist_100_y   = np.arange(1, len(dist_100)   + 1) / len(dist_100)
dist_1000_y  = np.arange(1, len(dist_1000)  + 1) / len(dist_1000)
dist_10000_y = np.arange(1, len(dist_10000) + 1) / len(dist_10000)

plt.figure(dpi=100)
plt.title('Функция распределения расстояний')
plt.plot(dist_100,   dist_100_y,   label='n=100')
plt.plot(dist_1000,  dist_1000_y,  label='n=1000')
plt.plot(dist_10000, dist_10000_y, label='n=10000')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(dpi=100)
plt.title('Плотность распределения расстояний')
plt.hist(dist_100,   density=True, label='n=100')
plt.hist(dist_1000,  density=True, label='n=1000')
plt.hist(dist_10000, density=True, label='n=10000')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
grid = np.linspace(0, 35, 200)

def ecdf(data, x):
    return np.searchsorted(np.sort(data), x, side='right') / len(data)

cdf_100   = ecdf(dist_100,   grid)
cdf_1000  = ecdf(dist_1000,  grid)
cdf_10000 = ecdf(dist_10000, grid)

plt.figure(dpi=100)
plt.title('Разница функций распределения (относительно n=10000)')
plt.plot(grid, cdf_100   - cdf_10000, label='n=100 − n=10000')
plt.plot(grid, cdf_1000  - cdf_10000, label='n=1000 − n=10000')
plt.legend()
plt.show()
```

```python
print(f"{'Выборка':<10} {'n':>7} {'Среднее':>12} {'Минимум':>12} {'Максимум':>12}")

for name, data in [('dist_100', dist_100), ('dist_1000', dist_1000), ('dist_10000', dist_10000)]:
    print(f"{name:<10} {len(data):>7} {np.average(data):>12.4f} {data.min():>12.4f} {data.max():>12.4f}")
```
